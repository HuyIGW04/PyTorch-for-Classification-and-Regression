### One Sample Code  
Minh họa từng bước, sử dụng một mẫu sample duy nhất.  
  
  
<img src="image.png" alt="image" width="400"/>


```python
import torch.nn as nn
import torch

# mẫu dữ liệu 1-D
x = torch.tensor([6.7])
y = torch.tensor([9.1])

# cài đặt linear - theta
linear = nn.Linear(1, 1)
linear.bias.data = torch.tensor([0.04])
linear.weight.data = torch.tensor([[-0.34]])

print(linear.bias)
print(linear.weight)
```

```python
# loss constructor
loss = nn.MSELoss()

# đưa linear - theta vào updating constructor
optimizer = torch.optim.SGD(linear.parameters(), lr=0.001)
```

```python
# đã chuẩn bị xong, bắt đầu quá trình training
losses = []
# compute predict
y_hat = linear(x)

# compute loss
loss_epoch = loss(y_hat, y)
losses.append(loss_epoch.item())

# compute gradient
# clear forward gradient
optimizer.zero_grad()
loss_epoch.backward()

# update weight
optimizer.step()
```

```python
y_hat = linear(x)
loss_epoch = loss(y_hat, y)
losses.append(loss_epoch.item())
```

```python
# kiểm tra loss xem có giảm chưa
losses
```
