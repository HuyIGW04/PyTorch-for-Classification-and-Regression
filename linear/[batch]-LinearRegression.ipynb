### HuyIGW04 - Batch Training - Linear Regression - PyTorch - M05W02

```python
import numpy as np
import torch.nn as nn
import torch
import matplotlib.pyplot as plt
```

```python
device = 'cuda'
data = np.genfromtxt('data.csv', delimiter=',')
X = torch.from_numpy(data[:, 0:1]).float().to(device)
y = torch.from_numpy(data[:, 1:]).float().to(device)
print(X.shape, y.shape)
```

```python
epochs = 100
losses = []

linear = nn.Linear(1, 1).to(device)
loss = nn.MSELoss()
optimizer = torch.optim.SGD(linear.parameters(), lr=0.001)

for epoch in range(epochs):
    y_hat = linear(X)
    loss_epoch = loss(y_hat, y)
    losses.append(loss_epoch.item())
    optimizer.zero_grad()
    loss_epoch.backward()
    optimizer.step()    
```

```python
plt.plot(losses)
```
